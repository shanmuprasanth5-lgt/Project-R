```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np
import matplotlib.pyplot as plt

# Define image parameters
img_height = 128
img_width = 128
batch_size = 32
```

Next, I will create a `tf.data.Dataset` from the image directories. I will combine the good and rotten onion paths into a single dataset, labeling them accordingly. I will use the `image_dataset_from_directory` utility from Keras to handle this.

```python
# Combine good_path and rotten_path for dataset creation
# Create a temporary directory structure for tf.keras.utils.image_dataset_from_directory
# This is a workaround since image_dataset_from_directory expects subdirectories for classes

base_dir = '/content/onion_dataset_combined'

# Clean up if previous run left directories
if os.path.exists(base_dir):
    import shutil
    shutil.rmtree(base_dir)

# Create class subdirectories
positive_dir = os.path.join(base_dir, 'good_onions')
negative_dir = os.path.join(base_dir, 'rotten_onions')
os.makedirs(positive_dir, exist_ok=True)
os.makedirs(negative_dir, exist_ok=True)

# Copy images to the new structure
# For 'good' onions
for filename in good_files:
    src = os.path.join(good_path, filename)
    dst = os.path.join(positive_dir, filename)
    os.link(src, dst) # Using os.link for efficiency if possible (creates hard link)

# For 'rotten' onions
for filename in rotten_files:
    src = os.path.join(rotten_path, filename)
    dst = os.path.join(negative_dir, filename)
    os.link(src, dst)

# Check the number of images copied
print(f"Copied {len(os.listdir(positive_dir))} good onion images.")
print(f"Copied {len(os.listdir(negative_dir))} rotten onion images.")

# Create the dataset
dataset = tf.keras.utils.image_dataset_from_directory(
    base_dir,
    labels='inferred',
    label_mode='binary',
    image_size=(img_height, img_width),
    interpolation='nearest',
    batch_size=batch_size,
    shuffle=True
)

# Split into training, validation, and test datasets
val_split = 0.2
test_split = 0.1

dataset_size = tf.data.experimental.cardinality(dataset).numpy() * batch_size

train_ds_size = int((1 - val_split - test_split) * dataset_size)
val_ds_size = int(val_split * dataset_size)
test_ds_size = int(test_split * dataset_size)

print(f"Dataset size: {dataset_size} images")
print(f"Train size (approx): {train_ds_size} images")
print(f"Validation size (approx): {val_ds_size} images")
print(f"Test size (approx): {test_ds_size} images")

# Shuffle and take subsets
dataset = dataset.shuffle(buffer_size=1000, reshuffle_each_iteration=True)

train_ds = dataset.take(train_ds_size // batch_size)
val_ds = dataset.skip(train_ds_size // batch_size).take(val_ds_size // batch_size)
test_ds = dataset.skip((train_ds_size + val_ds_size) // batch_size)

print(f"Number of training batches: {tf.data.experimental.cardinality(train_ds)}")
print(f"Number of validation batches: {tf.data.experimental.cardinality(val_ds)}")
print(f"Number of test batches: {tf.data.experimental.cardinality(test_ds)}")
```

Now, I'll normalize the pixel values to be between 0 and 1 and optimize performance using `tf.data` API methods.

```python
# The model's Rescaling layer will handle normalization. This preprocess function is no longer needed.

train_ds = train_ds.cache().prefetch(buffer_size=tf.data.AUTOTUNE)
val_ds = val_ds.cache().prefetch(buffer_size=tf.data.AUTOTUNE)
test_ds = test_ds.cache().prefetch(buffer_size=tf.data.AUTOTUNE)

print("Dataset caching and prefetching complete. Normalization will be handled by the model's Rescaling layer.")
```

Now, let's define the CNN model architecture. I'll use a sequential model with several convolutional and pooling layers, followed by dense layers.

```python
num_classes = 1 # Binary classification

model = keras.Sequential([
    layers.Rescaling(1./255, input_shape=(img_height, img_width, 3)),
    layers.Conv2D(16, 3, padding='same', activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(32, 3, padding='same', activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(64, 3, padding='same', activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(num_classes, activation='sigmoid') # Sigmoid for binary classification
])

model.summary()
```

### Fine-tuning with MobileNetV2

To improve the model's performance, especially on the 'Good' onion class, we will use a transfer learning approach with a pre-trained MobileNetV2 model. This involves:
1. Loading the MobileNetV2 model pre-trained on ImageNet.
2. Freezing its convolutional base to retain the learned features.
3. Adding a new classification head (dense layers) on top of the frozen base.
4. Compiling and training the new model.

```python
# Load the MobileNetV2 base model
base_model = tf.keras.applications.MobileNetV2(input_shape=(img_height, img_width, 3),
                                               include_top=False, # Don't include the ImageNet classifier
                                               weights='imagenet')

# Freeze the convolutional base
base_model.trainable = False

# Create a new model on top of the base
inputs = tf.keras.Input(shape=(img_height, img_width, 3))
x = base_model(inputs, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dense(128, activation='relu')(x)
outputs = layers.Dense(num_classes, activation='sigmoid')(x)

mobile_net_model = tf.keras.Model(inputs, outputs)

mobile_net_model.summary()
```

Now, let's compile and train the MobileNetV2-based model.

```python
mobile_net_model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
                     loss=tf.keras.losses.BinaryCrossentropy(),
                     metrics=['accuracy'])

# Train the new model
history_mobilenet = mobile_net_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=epochs
)

print("MobileNetV2 model training complete.")
```

To further improve the model's accuracy on the 'Good' onion class and potentially reduce false positives, we will implement data augmentation. This involves creating new training samples by applying random transformations to the existing images. We will then retrain the original CNN model using this augmented dataset.

```python
# Define data augmentation layers
data_augmentation = keras.Sequential(
  [
    layers.RandomFlip("horizontal", input_shape=(img_height, img_width, 3)),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
    layers.RandomContrast(0.2),
  ]
)

# Apply data augmentation to the training dataset
train_ds_augmented = train_ds.map(lambda x, y: (data_augmentation(x, training=True), y))

print("Data augmentation layers defined and applied to the training dataset.")
```

### Implementing Class Weighting and Retraining the Original CNN Model

Since the 'Good' onion class is under-represented, we will calculate class weights to assign a higher penalty to misclassifications of the minority class. This can help the model improve its performance on 'Good' onions. We will then retrain the original CNN model with these class weights.

```python
from sklearn.utils import class_weight

# Get all labels from the dataset to calculate class weights
all_labels_list = []
for _, labels_batch in dataset:
    # labels_batch.numpy() is expected to be shape (batch_size, 1) for binary labels.
    # Iterate through each item and append the scalar integer value.
    for label_array in labels_batch.numpy():
        all_labels_list.append(int(label_array[0]))

# Convert the list of scalar integers to a 1D NumPy array
all_labels = np.array(all_labels_list)

# Calculate class weights
class_weights = class_weight.compute_class_weight(
    class_weight='balanced',
    classes=np.unique(all_labels),
    y=all_labels
)

# Convert to a dictionary mapping class index to weight
class_weights_dict = {i: weight for i, weight in enumerate(class_weights)}

print("Calculated Class Weights:", class_weights_dict)
```

Now, let's redefine the original CNN model and train it with the calculated class weights.

```python
# Redefine the original CNN model (to ensure a fresh start)
model_weighted = keras.Sequential([
    layers.Rescaling(1./255, input_shape=(img_height, img_width, 3)),
    layers.Conv2D(16, 3, padding='same', activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(32, 3, padding='same', activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(64, 3, padding='same', activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(num_classes, activation='sigmoid')
])

model_weighted.compile(
    optimizer='adam',
    loss=tf.keras.losses.BinaryCrossentropy(),
    metrics=['accuracy']
)

model_weighted.summary()

print("Original CNN model redefined for weighted training.")
```

```python
# Train the model with class weights
history_weighted = model_weighted.fit(
    train_ds,
    validation_data=val_ds,
    epochs=epochs,
    class_weight=class_weights_dict
)

print("CNN model retraining with class weighting complete.")
```

Let's evaluate the class-weighted CNN model on the test dataset and visualize its performance with a confusion matrix.

```python
loss_weighted, accuracy_weighted = model_weighted.evaluate(test_ds)
print(f"Class-Weighted CNN Test Loss: {loss_weighted:.4f}")
print(f"Class-Weighted CNN Test Accuracy: {accuracy_weighted:.4f}")

# Get true labels and predictions for the class-weighted CNN model
y_true_weighted = []
y_pred_weighted = []

for images, labels in test_ds:
    y_true_weighted.extend(labels.numpy())
    predictions_weighted = model_weighted.predict(images)
    y_pred_weighted.extend(tf.round(predictions_weighted).numpy().flatten())

# Convert to numpy arrays
y_true_weighted = np.array(y_true_weighted)
y_pred_weighted = np.array(y_pred_weighted)

# Generate the confusion matrix for the class-weighted CNN
cm_weighted = confusion_matrix(y_true_weighted, y_pred_weighted)

# Plot the confusion matrix for the class-weighted CNN
plt.figure(figsize=(8, 6))
sns.heatmap(cm_weighted, annot=True, fmt='d', cmap='Blues', xticklabels=['Good', 'Rotten'], yticklabels=['Good', 'Rotten'])
plt.title('Confusion Matrix (Class-Weighted CNN)')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```

```python
acc_weighted = history_weighted.history['accuracy']
val_acc_weighted = history_weighted.history['val_accuracy']
loss_weighted_hist = history_weighted.history['loss']
val_loss_weighted = history_weighted.history['val_loss']

epochs_range = range(epochs)

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc_weighted, label='Weighted CNN Training Accuracy')
plt.plot(epochs_range, val_acc_weighted, label='Weighted CNN Validation Accuracy')
plt.legend(loc='lower right')
plt.title('Class-Weighted CNN Training and Validation Accuracy')

plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss_weighted_hist, label='Weighted CNN Training Loss')
plt.plot(epochs_range, val_loss_weighted, label='Weighted CNN Validation Loss')
plt.legend(loc='upper right')
plt.title('Class-Weighted CNN Training and Validation Loss')
plt.show()
```
